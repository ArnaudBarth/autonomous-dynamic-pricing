```python
# Peuple data/upsert/genres.csv (conversion / mobilité par genre) depuis le
# cache ChartMetric : le taux de conversion vient de l'engagement moyen
# (abonnés / auditeurs) des artistes du genre.
#
# Prérequis : avoir lancé le notebook 03 (ou scripts/fetch_chartmetric.py) pour
# remplir data/raw/chartmetric/artist_metadata.csv.

import os
import sys

sys.path.append(os.path.abspath("../env"))

from chartmetric import ChartmetricClient, CsvStore, fetch_artist, rebuild_genres_csv

store = CsvStore()
```

```python
# (optionnel) compléter le cache avec des artistes de référence par genre
REFERENCE_ARTISTS = {
    "Metal": ["Powerwolf", "Sabaton", "Electric Callboy", "Babymetal"],
    "Jazz": ["Ron Carter", "Wynton Marsalis", "Kamasi Washington"],
    "Classical": ["Hans Zimmer", "Joe Hisaishi"],
    # "Pop": ["Dua Lipa", "The Weeknd"],
    # "Indie": ["Phoebe Bridgers", "Fontaines D.C."],
}

client = ChartmetricClient()
for names in REFERENCE_ARTISTS.values():
    for name in names:
        fetch_artist(client, name, store=store)   # ne rappelle pas l'API si déjà en cache
```

```python
# ⚠️ genres.csv est normalement maintenu À LA MAIN : le proxy
# 'ratio abonnés/auditeurs' discrimine mal les genres quand les
# artistes vont de 5 k à 100 M d'auditeurs (tout finit dans le top palier).
# N'exécuter cette cellule que pour comparer / initialiser.
rebuild_genres_csv(store)
```
